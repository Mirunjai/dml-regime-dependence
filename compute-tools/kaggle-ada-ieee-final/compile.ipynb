```python
from pathlib import Path
import base64,io,zipfile,subprocess,tarfile,requests,json,hashlib,os,shutil
root=Path('/kaggle/working/ieee');root.mkdir()
zipfile.ZipFile(io.BytesIO(base64.b64decode(''))).extractall(root)
url='https://github.com/tectonic-typesetting/tectonic/releases/download/tectonic%400.17.0/tectonic-0.17.0-x86_64-unknown-linux-musl.tar.gz'
r=requests.get(url,timeout=120);r.raise_for_status()
with tarfile.open(fileobj=io.BytesIO(r.content),mode='r:gz') as t:t.extractall(root,filter='data')
result=subprocess.run([str(root/'tectonic'),'--keep-logs','ADA_IEEE_Report.tex'],cwd=root,env={**os.environ,"XDG_CACHE_HOME":str(root/"texcache")},text=True,stdout=subprocess.PIPE,stderr=subprocess.STDOUT,timeout=360)
(root/'compile.log').write_text(result.stdout)
shutil.make_archive("/kaggle/working/tectonic_cache","zip",root_dir=root/"texcache")
print(result.stdout[-6000:]);assert result.returncode==0
Path('/kaggle/working/compilation.json').write_text(json.dumps({'returncode':result.returncode,'tectonic_binary_sha256':hashlib.sha256((root/'tectonic').read_bytes()).hexdigest()}))
```
