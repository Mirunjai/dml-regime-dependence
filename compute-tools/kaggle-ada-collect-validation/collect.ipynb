# Frozen exploratory archive collection
Integrity-check the two completed CPU kernels; no new simulations.

```python
from pathlib import Path
import json,hashlib,shutil
roots=[p.parent for p in Path('/kaggle/input').rglob('completion.json') if p.parent.name in ('validation_tau0','validation_tau1')]
assert {p.name for p in roots}=={'validation_tau0','validation_tau1'}, str(roots)
records=[]
for root in sorted(roots):
    completion=json.loads((root/'completion.json').read_text())
    assert completion['complete'] and completion['blocks']==1600
    for name,digest in completion['files_sha256'].items():
        assert hashlib.sha256((root/name).read_bytes()).hexdigest()==digest
    archive=Path(shutil.make_archive('/kaggle/working/'+root.name,'zip',root_dir=root))
    records.append({'input':str(root),'archive':archive.name,'sha256':hashlib.sha256(archive.read_bytes()).hexdigest(),'blocks':completion['blocks']})
Path('/kaggle/working/collection.json').write_text(json.dumps(records,indent=2))
print(records)
```
